```python
df["Revenue"] = df["Quantity"] * df["Price"]
```

```python
df.head()
```

```python
df_clean = df.copy()
```

```python
df_clean.drop_duplicates(inplace=True)

print("Rows after removing duplicates:", df_clean.shape[0])
```

```python
df_clean = df_clean.dropna(subset=["Description"])
```

```python
df_clean = df_clean[df_clean["Quantity"] > 0]
```

```python
# Remove rows with a zero or negative price (free samples, write-offs, data errors)
df_clean = df_clean[df_clean["Price"] > 0]

print("Rows after removing invalid prices:", df_clean.shape[0])
```

```python
# Remove cancelled invoices (invoice numbers starting with "C")
df_clean = df_clean[~df_clean["Invoice"].astype(str).str.startswith("C")]

print("Rows after removing cancellations:", df_clean.shape[0])
```

### Non-product lines

Some stock codes are not products: postage, manual adjustments, marketplace fees, bank charges, discounts, samples, test items and gift vouchers. Left in, "Manual" shows up as the top-selling "product", so they are removed before any product or revenue analysis.

```python
non_product_codes = [
    "M", "POST", "DOT", "C2", "AMAZONFEE", "BANK CHARGES", "D",
    "ADJUST", "ADJUST2", "S", "B", "PADS", "CRUK", "TEST001", "TEST002",
]

stock = df_clean["StockCode"].astype(str).str.upper()
is_non_product = stock.isin(non_product_codes) | stock.str.startswith("GIFT_")

print("Non-product rows:", is_non_product.sum())
print(
    df_clean[is_non_product]
    .groupby("Description")["Revenue"].sum()
    .sort_values(ascending=False)
    .head(10)
)

df_clean = df_clean[~is_non_product]

print("Rows after removing non-product lines:", df_clean.shape[0])
```
